```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
path = "victor (1).xlsx"
dataraw = pd.read_excel(path)
print("=== TABEL DATA ===")
print(dataraw)
```

```python
urutan = [g for g in ["A", "AB", "B", "BC", "C", "D", "E", "F"]
          if g in dataraw["Grade"].unique()]
datafrq = pd.crosstab(index=dataraw["Grade"], columns="Frekuensi",
                      colnames=["Keterangan"]).reindex(urutan)
datafrq["Persentase (%)"] = (datafrq["Frekuensi"] / datafrq["Frekuensi"].sum() * 100).round(1)
print("\n=== TABEL FREKUENSI ===")
print(datafrq)
print("Total :", datafrq["Frekuensi"].sum(), "mahasiswa")
```

```python
warna = ["#1F2A5C", "#3D5A9E", "#6F8FD1", "#F2A541", "#E4572E", "#8C2F39"]
```

```python
# Grafik Garis
ax = datafrq["Frekuensi"].plot(marker="o", color="#1F2A5C", linewidth=2, figsize=(7, 4.5))
ax.set_title("Grafik Garis Frekuensi Grade Mahasiswa")
ax.set_xlabel("Grade")
ax.set_ylabel("Jumlah Mahasiswa")
ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("grafik_garis.png", dpi=150)
plt.show()
plt.close()
```

```python
# Grafik Batang
ax = datafrq["Frekuensi"].plot(kind="bar", color=warna[:len(datafrq)], figsize=(7, 4.5))
ax.set_title("Grafik Batang Frekuensi Grade Mahasiswa")
ax.set_xlabel("Grade")
ax.set_ylabel("Jumlah Mahasiswa")
ax.bar_label(ax.containers[0])       # tampilkan angka di atas batang
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("grafik_batang.png", dpi=150)
plt.show()
plt.close()
```

```python
# Pie Chart
ax = datafrq.plot(kind="pie", y="Frekuensi", autopct="%1.1f%%", legend=False,
                  colors=warna[:len(datafrq)], startangle=90, counterclock=False,
                  figsize=(6, 6))
ax.set_title("Pie Chart Persentase Grade Mahasiswa")
ax.set_ylabel("")
plt.tight_layout()
plt.savefig("pie_chart.png", dpi=150)
plt.show()
plt.close()
```

```python
dataraw["Nilai"] = pd.to_numeric(dataraw["Nilai"], errors="coerce")
dt = dataraw["Nilai"]

stats = dt.describe()
stats["Standard Error"] = dt.sem()
stats["Median"] = dt.median()
stats["Mode"] = dt.mode().iloc[0]
stats["Variance"] = dt.var()
stats["Range"] = dt.max() - dt.min()
stats["Skewness"] = dt.skew()
stats["Kurtosis"] = dt.kurtosis()
print("\n=== STATISTIKA DESKRIPTIF NILAI AKHIR ===")
print(stats)
```
